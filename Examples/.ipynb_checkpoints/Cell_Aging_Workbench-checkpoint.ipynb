```python
from scipy.stats import t as sci_t
import numpy as np
import pandas as pd
def t_value_eqn(x1, s1, n1, x2, s2, n2):
    '''
    Performs a two-tailed t-test. 
    '''
    return (x1 - x2)/(np.sqrt(((s1**2)/n1) + ((s2**2 )/ n2)))

def compare_data(mean1, std1, len1, mean2, std2, len2):
    '''
    Evaluates 2-tailed t and p values.
    '''
    t_val = t_value_eqn(mean1, std1, len1, mean2, std2, len2)
    p_val = sci_t.sf(abs(t_val), len1+len2-2)*2
    return abs(t_val), p_val
```

```python
def cell_SOH_loss_by_DB(dq, c, dt):
    '''USe the DB equation to determine cell capacity decay'''
    B = .0001
    k = .0005
    c=abs(c)
    dq=abs(dq)

    return B*(dq/1)*c*dt - k*(dt)**.5
```

```python
def cell_SOH_loss_by_DB(dq, c, dt):
    #LFP Param for B: .00021
    #NMC param for B: 
    '''USe the DB equation to determine cell capacity decay'''
    B = .00021
    k = .0001
    Q_0 = .120 #Ah
    c=abs(c)
    dq=abs(dq)

    return -(B*c*dq/2 )

import matplotlib.pyplot as plt


fig, ax = plt.subplots(1, 1, figsize = (10, 10), dpi = 100)
ax.set_ylim(0, .140)
vals=[]
t = []
for i in range(1,2000):
    
    vals.append(cell_SOH_loss_by_DB(.120, 1, i*60))
    t.append(i)

ax.plot(np.asarray(t), np.asarray(vals).cumsum()+.120)
vals=[]
t = []
for i in range(1,667):
    
    vals.append(cell_SOH_loss_by_DB(.120, 3, i*60))
    t.append(i)

ax.plot(np.asarray(t)*3, np.asarray(vals).cumsum()+.100)
vals=[]
t = []
for i in range(1,400):
    
    vals.append(cell_SOH_loss_by_DB(.120, 5, i*60))
    t.append(i)

ax.plot(np.asarray(t)*5, np.asarray(vals).cumsum()+.060)

ax.grid()
ax.set_xlabel('Cycle #')
ax.set_ylabel('Capacity [Ah]')
ax.set_title("LFP Aging")
ax.legend(['C', '3C', '5C'])
```

```python
linedata = {}
lines = None
with open('/media/sebastian/Sindri/Acceleration_Profile_Test_Results_2') as f:
    lines = f.readlines()
current_token = ' '
for i in range(len(lines)):
    line = lines[i]
    if '----' in line:
        token = line.split('.')[0].split('----')[1]
        current_token = token
        linedata[current_token] = {'A_r^2':[],'V_r^2':[],'MC_vs_GPS':[],'MC_vs_EX':[]}
    else:
        if 'A r^2' in line:
            linedata[current_token]['A_r^2'].append(float(line.split(': ')[1].split(',')[0]))
            linedata[current_token]['V_r^2'].append(float(line.split(': ')[2]))
        if 'GPS' in line:
            linedata[current_token]['MC_vs_GPS'].append(float(line.split(', ')[1].split(')')[0]))
        if 'EX' in line:
            linedata[current_token]['MC_vs_EX'].append(float(line.split(', ')[1].split(')')[0]))
                

linedataframe = pd.DataFrame(linedata)
#print(linedataframe)
#linedataframe['Route'] = linedataframe['Braunschweig'].apply(lambda x: [45, 70, 8])
#linedataframe = linedataframe.explode(list(linedataframe.columns))

'''
for route in [45, 70, 8]:
    print("------------{}-----------".format(route))
    route_data = linedataframe.groupby('Route').get_group(route).drop('Route', axis=1).T                   #.T.explode(['V_r^2', 'A_r^2', 'MC_vs_GPS', 'MC_vs_EX'])
    best_A_fits = route_data[(route_data['A_r^2'] > 0) & (route_data['MC_vs_GPS'] > 0) ].sort_values('V_r^2', ascending = False)[:6]
    print(best_A_fits)
'''
seq_means_df = linedataframe.applymap(lambda x: np.asarray(x).mean()).T
print(seq_means_df[seq_means_df['MC_vs_GPS'] > 0.00].sort_values('A_r^2', ascending = False))
```
